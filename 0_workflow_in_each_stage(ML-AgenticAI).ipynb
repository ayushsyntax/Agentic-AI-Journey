```python
# tabular ML workflow
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

df = pd.read_csv("data.csv")
X = df.drop("target", axis=1)
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = RandomForestClassifier()
model.fit(X_train, y_train)

pred = model.predict(X_test)
print(accuracy_score(y_test, pred))
```

Key points: leakage prevention, scaling after split, simple baselines first.

LangGraph

```python
# minimal langgraph agent
from langgraph.graph import StateGraph, END
from typing import TypedDict
from langchain.chat_models import ChatOpenAI

class State(TypedDict):
    question: str
    answer: str

llm = ChatOpenAI()

def node_llm(state:State):
    return {"answer": llm.invoke(state["question"]).content}

g = StateGraph(State)
g.add_node("llm", node_llm)
g.set_entry_point("llm")
g.add_edge("llm", END)

app = g.compile()
print(app.invoke({"question":"2+2"}))
```

Key points: explicit state; nodes + edges; loops/branches scale agents.

## 3. Outlier treatment

Goal: prevent scale distortion.

```python
import numpy as np

q1,q3 = df["income"].quantile([0.25,0.75])
iqr = q3-q1
low,high = q1-1.5*iqr, q3+1.5*iqr
df["income"] = df["income"].clip(low,high)

df["income_log"] = np.log1p(df["income"])
```

# Universal feature-engineering protocol

```
1. Identify data modality
2. Normalize raw scale/format
3. Encode structure (category/space/time/sequence)
4. Inject invariances (augmentations or transforms)
5. Add domain composites
6. Remove noise/redundancy
7. Package into pipeline/state
```

Everything below is just modality-specific realization of this.

```
missing → outliers → scale → encode → interactions → select
```

```python
num = df[num_cols].copy()
cat = df[cat_cols].copy()

num = num.fillna(num.median())
num = num.clip(num.quantile(.01), num.quantile(.99), axis=1)
num = (num - num.mean())/num.std()

for c in cat:
    freq = cat[c].value_counts(normalize=True)
    num[c+"_freq"] = cat[c].map(freq)

num["ratio1"] = num["income"]/(num["age"]+1)

X = num
```

Rules that remove guesswork:

• trees → skip scaling
• linear/ANN → always scale
• high-cardinality cat → frequency
• low-cardinality cat → one-hot
